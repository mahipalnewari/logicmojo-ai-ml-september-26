Scenario: You are building a simple student result management system.
Write a Python program that performs the following tasks:
1. Ask the user to enter marks for 5 students using a while loop.
2. If the entered marks are outside the range of 0–100, display an invalid marks message and skip that entry using continue. The user must still enter marks until 5 valid entries have been collected.
3. Use a for loop to process the collected marks.
4. If marks are 90 or above, print Grade A.
5. If marks are between 75 and 89, print Grade B.
6. If marks are between 60 and 74, print Grade C.
7. If marks are between 40 and 59, print Pass.
8. Otherwise, print Fail.
9. If marks are exactly 0, use break to stop processing further students.
10. Include a pass statement as a placeholder for a future feature, such as saving student results to a file.
11. Finally, print how many valid marks were collected.

```python
# Student Result Management System

count = 0

# WHILE LOOP: Collect marks for 5 students
while count < 5:

    marks = int(input("Enter marks (0-100): "))

    # IF: Validate marks
    if marks < 0 or marks > 100:
        print("Invalid marks. Try again.")
        continue

    count = count + 1

    # BREAK: Stop when marks are zero
    if marks == 0:
        print("Zero marks entered. Stopping result entry.")
        break

    # IF-ELIF-ELSE: Assign grade
    if marks >= 90:
        print("Grade A")
    elif marks >= 75:
        print("Grade B")
    elif marks >= 60:
        print("Grade C")
    elif marks >= 40:
        print("Grade D")
    else:
        print("Grade F")

    # IF-ELSE: Pass or fail
    if marks >= 40:
        print("Result: Pass")
    else:
        print("Result: Fail")

    # NESTED IF: Scholarship eligibility
    if marks >= 75:
        if marks >= 90:
            print("Eligible for full scholarship")
        else:
            print("Eligible for partial scholarship")
    else:
        print("Not eligible for scholarship")

    # PASS: Placeholder for future feature
    if marks == 50:
        pass

    print("--------------------")


# FOR LOOP: Countdown demonstration
print("Countdown:")

for i in range(1, 6):
    print(i)

print("Valid entries processed:", count)
```

Create a BMI (Body Mass Index) Calculator using Python. Your program should demonstrate conditional statements, loops and loop-control statements without using lists or membership operators.
Requirements
1. Use a while loop to allow the user to calculate BMI for multiple people.
2. Ask the user to enter their weight in kilograms and height in metres.
3. If the weight or height is zero or negative, display an invalid-input message and use continue to ask again.
4. Calculate BMI using the formula:
   BMI = weight (kg) ÷ height (m)²
5. Use if-elif-else to classify BMI:
   - Below 18.5: Underweight
   - 18.5 to below 25: Normal weight
   - 25 to below 30: Overweight
   - 30 or above: Obesity range
6. Use a nested if to provide an additional message for people in the normal BMI range.
7. Use if-else to check whether BMI is below 25.
8. Use break to stop the program when the user enters 0 as their weight.
9. Use a for loop to display a countdown from 3 to 1 before exiting.
10. Use pass as a placeholder for a future feature.

```python
# BMI Calculator

count = 0

while True:

    weight = float(input("Enter weight in kg (0 to exit): "))

    # 1. BREAK: Exit when weight is 0
    if weight == 0:
        print("Exiting BMI Calculator...")
        break

    # 2. IF: Validate weight
    if weight < 0:
        print("Invalid weight! Please enter a positive value.")
        continue

    height = float(input("Enter height in metres: "))

    # 3. CONTINUE: Skip invalid height
    if height <= 0:
        print("Invalid height! Please try again.")
        continue

    # 4. Calculate BMI
    bmi = weight / (height ** 2)

    print("Your BMI is:", round(bmi, 2))

    # 5. IF-ELIF-ELSE: BMI classification
    if bmi < 18.5:
        print("Category: Underweight")

    elif bmi < 25:
        print("Category: Normal weight")

        # 6. NESTED IF: Additional message
        if bmi >= 20:
            print("Your BMI is in the upper part of the normal range.")
        else:
            print("Your BMI is in the lower part of the normal range.")

    elif bmi < 30:
        print("Category: Overweight")

    else:
        print("Category: Obesity range")

    # 7. IF-ELSE: Check whether BMI is below 25
    if bmi < 25:
        print("BMI is below 25.")
    else:
        print("BMI is 25 or above.")

    # 8. PASS: Placeholder for future feature
    if bmi == 25:
        pass

    count = count + 1

    print("-------------------------")

# 9. FOR LOOP: Countdown after exiting
for i in range(3, 0, -1):
    print(i)

print("Total calculations:", count)
```

```python
# ==========================================
# PYTHON SET: ALL METHODS AND OPERATIONS
# ==========================================

# 1. Creating a set
numbers = {10, 20, 30, 40, 50}
print(numbers)

# 2. Checking the type
print(type(numbers))

# 3. Creating an empty set
empty_set = set()
print(empty_set)

# 4. Creating a set from a list
numbers_from_list = set([10, 20, 20, 30, 30, 40])
print(numbers_from_list)

# 5. Finding the length
print(len(numbers))

# 6. Adding one element using add()
numbers.add(60)
print(numbers)

# 7. Adding an existing element
numbers.add(20)
print(numbers)

# 8. Adding multiple elements using update()
numbers.update([70, 80, 90])
print(numbers)

# 9. Updating from another set
numbers.update({100, 110})
print(numbers)

# 10. Removing an element using remove()
numbers.remove(20)
print(numbers)

# 11. Removing an element using discard()
numbers.discard(30)
print(numbers)

# 12. Discarding a nonexistent element
numbers.discard(999)
print(numbers)

# 13. Removing an arbitrary element using pop()
removed_value = numbers.pop()
print("Removed:", removed_value)
print(numbers)

# 14. Checking membership using in
print(40 in numbers)

# 15. Checking membership using not in
print(999 not in numbers)

# 16. Creating two sets for set operations
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}

# 17. Union using the | operator
print(A | B)

# 18. Union using union()
print(A.union(B))

# 19. Intersection using the & operator
print(A & B)

# 20. Intersection using intersection()
print(A.intersection(B))

# 21. Difference using the - operator
print(A - B)

# 22. Difference using difference()
print(A.difference(B))

# 23. Reverse difference
print(B - A)

# 24. Symmetric difference using ^ 
print(A ^ B)

# 25. Symmetric difference using symmetric_difference()
print(A.symmetric_difference(B))

# 26. Checking whether A is a subset of B
print(A.issubset(B))

# 27. Checking whether one set is a subset of another
print({1, 2}.issubset(A))

# 28. Checking whether A is a superset of another set
print(A.issuperset({1, 2}))

# 29. Checking whether two sets are disjoint
print({1, 2}.isdisjoint({3, 4}))

# 30. Updating a set using intersection_update()
C = {1, 2, 3, 4}
D = {3, 4, 5, 6}
C.intersection_update(D)
print(C)

# 31. Updating a set using difference_update()
C = {1, 2, 3, 4}
D = {3, 4, 5, 6}
C.difference_update(D)
print(C)

# 32. Updating a set using symmetric_difference_update()
C = {1, 2, 3, 4}
D = {3, 4, 5, 6}
C.symmetric_difference_update(D)
print(C)

# 33. Updating a set using |=
C = {1, 2}
C |= {3, 4}
print(C)

# 34. Updating a set using &=
C = {1, 2, 3}
C &= {2, 3, 4}
print(C)

# 35. Updating a set using -=
C = {1, 2, 3}
C -= {2}
print(C)

# 36. Updating a set using ^=
C = {1, 2, 3}
C ^= {3, 4}
print(C)

# 37. Copying a set using copy()
original = {10, 20, 30}
copied = original.copy()
copied.add(40)
print(original)
print(copied)

# 38. Converting a set into a list
numbers_list = list(A)
print(numbers_list)

# 39. Converting a set into a tuple
numbers_tuple = tuple(A)
print(numbers_tuple)

# 40. Removing duplicates from a list using set()
duplicate_list = [1, 2, 2, 3, 3, 4, 4, 5]
unique_values = set(duplicate_list)
print(unique_values)

# 41. Iterating through a set
for item in A:
    print(item)

# 42. Finding minimum and maximum
print(min(A))
print(max(A))

# 43. Calculating the sum
print(sum(A))

# 44. Comparing sets for equality
print({1, 2, 3} == {3, 2, 1})

# 45. Clearing all elements using clear()
C = {10, 20, 30}
C.clear()
print(C)

# 46. Viewing available set methods
print(dir(A))
```

```python
# ==========================================
# PYTHON DICTIONARY: ALL METHODS AND OPERATIONS
# ==========================================

# 1. Creating a dictionary
student = {
    "name": "Aman",
    "age": 21,
    "marks": 85,
    "city": "Jaipur"
}
print(student)

# 2. Checking the type
print(type(student))

# 3. Creating an empty dictionary
empty_dict = {}
print(empty_dict)

# 4. Creating a dictionary using dict()
student2 = dict(name="Priya", age=22, marks=92)
print(student2)

# 5. Creating a dictionary using zip()
names = ["Aman", "Priya", "Rahul"]
marks = [85, 92, 78]
student_marks = dict(zip(names, marks))
print(student_marks)

# 6. Accessing a value using its key
print(student["name"])
print(student["marks"])

# 7. Accessing a value using get()
print(student.get("name"))

# 8. Getting a missing key with get()
print(student.get("grade"))

# 9. Providing a default value using get()
print(student.get("grade", "Not Available"))

# 10. Adding a new key-value pair
student["grade"] = "A"
print(student)

# 11. Updating an existing value
student["marks"] = 95
print(student)

# 12. Updating multiple entries using update()
student.update({"age": 22, "city": "Delhi"})
print(student)

# 13. Using setdefault() for a missing key
student.setdefault("country", "India")
print(student)

# 14. Using setdefault() for an existing key
student.setdefault("marks", 0)
print(student)

# 15. Finding the number of key-value pairs
print(len(student))

# 16. Checking whether a key exists
print("name" in student)

# 17. Checking whether a key does not exist
print("salary" not in student)

# 18. Checking whether a value exists
print("Aman" in student.values())

# 19. Getting all keys using keys()
print(student.keys())

# 20. Getting all values using values()
print(student.values())

# 21. Getting all key-value pairs using items()
print(student.items())

# 22. Converting keys into a list
print(list(student.keys()))

# 23. Converting values into a list
print(list(student.values()))

# 24. Converting items into a list
print(list(student.items()))

# 25. Iterating through dictionary keys
for key in student:
    print(key)

# 26. Iterating through dictionary values
for value in student.values():
    print(value)

# 27. Iterating through key-value pairs
for key, value in student.items():
    print(key, ":", value)

# 28. Removing an entry using pop()
removed_value = student.pop("age")
print(removed_value)
print(student)

# 29. Using pop() with a default value
print(student.pop("salary", "Not Found"))

# 30. Removing the last inserted pair using popitem()
removed_item = student.popitem()
print(removed_item)
print(student)

# 31. Deleting an entry using del
del student["city"]
print(student)

# 32. Copying a dictionary using copy()
original = {"name": "Aman", "marks": 85}
copied = original.copy()
copied["marks"] = 95
print(original)
print(copied)

# 33. Creating a dictionary using fromkeys()
subjects = ["Python", "SQL", "Excel"]
subject_status = dict.fromkeys(subjects, "Pending")
print(subject_status)

# 34. Creating a dictionary using dictionary comprehension
squares = {x: x**2 for x in range(1, 6)}
print(squares)

# 35. Dictionary comprehension with a condition
marks = {"Aman": 85, "Priya": 92, "Rahul": 65, "Neha": 78}
high_scorers = {
    name: score
    for name, score in marks.items()
    if score >= 80
}
print(high_scorers)

# 36. Nested dictionary
students = {
    "S101": {"name": "Aman", "marks": 85},
    "S102": {"name": "Priya", "marks": 92}
}
print(students)

# 37. Accessing a nested dictionary value
print(students["S102"]["marks"])

# 38. Updating a nested dictionary value
students["S101"]["marks"] = 90
print(students)

# 39. Merging two dictionaries using update()
dict1 = {"a": 1, "b": 2}
dict2 = {"b": 20, "c": 3}
dict1.update(dict2)
print(dict1)

# 40. Merging dictionaries using the | operator (Python 3.9+)
dict1 = {"a": 1, "b": 2}
dict2 = {"b": 20, "c": 3}
merged = dict1 | dict2
print(merged)

# 41. Creating a dictionary from a list of tuples
pairs = [("a", 1), ("b", 2), ("c", 3)]
result = dict(pairs)
print(result)

# 42. Getting keys in insertion order
print(list(student.keys()))

# 43. Clearing all entries using clear()
student.clear()
print(student)

# 44. Viewing available dictionary methods
example = {"a": 1, "b": 2}
print(dir(example))
```
